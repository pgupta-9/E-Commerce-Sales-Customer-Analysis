```python
# Convert date columns in Orders dataset to datetime format

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders[column] = pd.to_datetime(orders[column], errors="coerce")

print("Date columns converted successfully")
```

```python
orders[date_columns].dtypes
```

```python
# Convert date columns in other datasets

order_items["shipping_limit_date"] = pd.to_datetime(
    order_items["shipping_limit_date"], errors="coerce"
)

reviews["review_creation_date"] = pd.to_datetime(
    reviews["review_creation_date"], errors="coerce"
)

reviews["review_answer_timestamp"] = pd.to_datetime(
    reviews["review_answer_timestamp"], errors="coerce"
)

print("Other date columns converted successfully")
```

```python
print(order_items["shipping_limit_date"].dtype)
print(reviews["review_creation_date"].dtype)
print(reviews["review_answer_timestamp"].dtype)
```

```python
# Handle missing values in product dataset

# Missing product categories ko "Unknown" se replace karna
products["product_category_name"] = products["product_category_name"].fillna("Unknown")

# Numeric missing values ko median se replace karna
numeric_columns = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

for column in numeric_columns:
    products[column] = products[column].fillna(products[column].median())

print("Product missing values handled successfully")
```

```python
# Handle missing values in product data

products["product_category_name"] = products["product_category_name"].fillna("Unknown")

for col in ["product_weight_g", "product_length_cm",
            "product_height_cm", "product_width_cm"]:
    products[col] = products[col].fillna(products[col].median())

print("Data cleaning completed successfully")
```

```python
# Create useful columns for analysis

orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.days

orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.days

print("Analysis columns created successfully")
```

```python
print("ORDER ITEMS STATISTICS")
display(order_items[["price", "freight_value"]].describe())

print("ORDERS BY STATUS")
display(orders["order_status"].value_counts())

print("TOP 10 CUSTOMER STATES")
display(customers["customer_state"].value_counts().head(10))

print("TOP 10 PRODUCT CATEGORIES")
display(products["product_category_name"].value_counts().head(10))
```

```python
# Monthly sales analysis

order_items["order_id"].nunique()
```

```python
# Merge order items with order dates

sales_data = order_items.merge(
    orders[["order_id", "order_purchase_timestamp"]],
    on="order_id",
    how="left"
)

# Create month column
sales_data["month"] = sales_data["order_purchase_timestamp"].dt.to_period("M")

# Monthly revenue
monthly_sales = sales_data.groupby("month")["price"].sum()

print("Monthly Sales:")
display(monthly_sales)
```

```python
# Delivery analysis

print("Average Delivery Days:",
      round(orders["delivery_days"].mean(), 2))

print("Maximum Delivery Days:",
      orders["delivery_days"].max())

print("Average Delivery Delay:",
      round(orders["delivery_delay_days"].mean(), 2))
```

```python
# Correlation analysis

correlation = order_items[["price", "freight_value"]].corr()

print("Correlation Matrix:")
display(correlation)
```

```python
# Outlier detection using IQR

Q1 = order_items["price"].quantile(0.25)
Q3 = order_items["price"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = order_items[
    (order_items["price"] < lower_limit) |
    (order_items["price"] > upper_limit)
]

print("Number of price outliers:", len(outliers))
print("Lower Limit:", round(lower_limit, 2))
print("Upper Limit:", round(upper_limit, 2))
```
